Scaled Dot-Product Attention

```python

import numpy as np

def scaled_dot_product_attention(Q, K, V):
    # Step 1: Calculate dot-product attention scores
    scores = np.matmul(Q, K.T)

    # Step 2: Scale by square root of key dimension
    d = K.shape[-1]
    scaled_scores = scores / np.sqrt(d)

    # Step 3: Apply softmax
    exp_scores = np.exp(
        scaled_scores - np.max(scaled_scores, axis=-1, keepdims=True)
    )
    attention_weights = exp_scores / np.sum(
        exp_scores, axis=-1, keepdims=True
    )

    # Step 4: Calculate final output
    output = np.matmul(attention_weights, V)

    return attention_weights, output


# Test inputs
Q = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
K = np.array([[1, 0, 1, 0], [0, 1, 0, 1]])
V = np.array([[1, 2, 3, 4], [5, 6, 7, 8]])

# Run attention function
weights, output = scaled_dot_product_attention(Q, K, V)

print("Attention Weights Matrix:")
print(np.round(weights, 3))

print("\nFinal Output Matrix:")
print(np.round(output, 3))
```

Dividing by √d prevents attention scores from becoming too large as the key dimension increases. Large scores can cause the softmax function to produce extremely uneven weights, resulting in very small gradients. Scaling helps maintain stable gradients and improves model training.

Self-attention allows each word to focus on other relevant words in the sentence, regardless of their positions. It assigns attention weights based on relationships between words, helping the model understand context, meaning, and long-range dependencies. For example, in "The student finished her assignment because she studied," self-attention can connect "she" to "student."
